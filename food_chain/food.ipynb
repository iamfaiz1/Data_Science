```python

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np

df = pd.read_csv("FoodChain_ComparativeDataset.csv")
df 

print("avg: ", df["Delivery_Time1"].mean())
print("avg: ", df["Delivery_Time2"].mean())

# how much fluctuation is there in the data
print("std: ",df["Delivery_Time1"].std())
print("std: ",df["Delivery_Time2"].std())
```
