```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# load dataset
df = sns.load_dataset('tips')

# display first few rows of the dataset
print("Simple dataSet:")

with open('output.txt', 'w') as f:
    df.to_string(f)

# df.describe()
# df.info()
```

```python

# scatter plot- visualising statistical relationsjop
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='total_bill', y='tip')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python

# scatter plot- visualising statistical relationsjop
plt.figure(figsize=(10, 6))
# hue parameter adds color coding based on a categorical variable onlu one 
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
# scatter plot- visualising statistical relationsjop
plt.figure(figsize=(10, 6))
# hue parameter adds color coding based on a categorical variable onlu one 
# sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex', style='smoker')
sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex', size='smoker')
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

# analyse male and female visiters on day basis

```python
plt.figure(figsize=(10, 6))
# hue parameter adds color coding based on a categorical variable onlu one 

sns.countplot(data=df, x='day', hue='sex' )
plt.title('visits of male and female on day basis')
plt.show()
```

```python
df.groupby(['day']).size()
```

```python
plt.figure(figsize=(10, 6))
# hue parameter adds color coding based on a categorical variable onlu one 

sns.countplot(data=df, x='day', hue='time' )
plt.title('visits of male and female on day basis')
plt.show()
```

```python
df.groupby(['time']).size()
```

# box whiskers plot

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='day', y ='total_bill', hue='sex')
plt.title('box plot: total bill distribution by day and sex')
plt.show()
# box whiskers plot
```

```python
plt.figure(figsize=(10, 6))

sns.swarmplot(data= df, x= 'day', y='total_bill', hue='sex')
plt.title('swarm plot: total bill distribution by day and')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
# heat map
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='Reds')
plt.title('Heatmap of Numerical Feature Correlations')
plt.show()
```

```python


plt.figure(figsize=(10, 6))
# heat map
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='Greens')
plt.title('Heatmap of Numerical Feature Correlations')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
# heat map
sns.heatmap(df.select_dtypes(include='number').corr(), annot=False, cmap='Greens')
plt.title('Heatmap of Numerical Feature Correlations')
plt.show()
```
