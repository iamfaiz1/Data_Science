```python
import seaborn as sns
import matplotlib.pyplot as plt
mtcars = sns.load_dataset("mpg").dropna()
print(mtcars.head())
print(mtcars.describe())

# mpg : mileage per gallon

```

```python

mean = mtcars["mpg"].mean()
std = mtcars["mpg"].std()

print("mean: ", mean)
print("std: ", std)
```

```python
# plotting histogram with std ranges
data = mtcars["mpg"]
sns.histplot(data, kde=True, bins=10, color='green')

col = "mpg"
for i in range(1, 4):
    plt.axvline(mean + i*std, color="red", linestyle="--")
    plt.axvline(mean - i*std, color="red", linestyle="--")

plt.title(f"Empiriclal rule Demonstration - {col}")
plt.show()
```

```python
from scipy.stats import skew, kurtosis

for col in ["horsepower", "weight", "mpg"]:
    print(f"{col} skewness: {skew(mtcars[col]):.2f}")
    sns.histplot(mtcars[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# from scipy.stats import skew, kurtosis

for col in ["horsepower", "weight", "mpg"]:
    print(f"{col} Kurtosis: {kurtosis(mtcars[col]):.2f}")
    sns.histplot(mtcars[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```
